```python
from dyadic_piano_env import DyadicPianoEnv
from dyadic_q_agent import DyadicQAgent


env = DyadicPianoEnv(seed=42)

agent_A1 = DyadicQAgent()
agent_A2 = DyadicQAgent()

episodes = 5000
steps = 100

for ep in range(episodes):

    state = env.reset()

    total_reward_A1 = 0
    total_reward_A2 = 0

    for t in range(steps):

        action_A1 = agent_A1.select_action(state)
        action_A2 = agent_A2.select_action(state)

        next_state, reward_A1, reward_A2 = env.step(
            action_A1,
            action_A2
        )

        agent_A1.update(
            state,
            action_A1,
            reward_A1,
            next_state
        )

        agent_A2.update(
            state,
            action_A2,
            reward_A2,
            next_state
        )

        state = next_state

        total_reward_A1 += reward_A1
        total_reward_A2 += reward_A2

    if ep % 500 == 0:

        print(
            f"Episode {ep} | "
            f"A1 reward: {total_reward_A1:.3f} | "
            f"A2 reward: {total_reward_A2:.3f}"
        )
```
